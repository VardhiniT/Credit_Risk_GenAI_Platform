```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
df = pd.read_csv("../data/raw/FreddieMac.csv")

print(df.shape)
df.head()
```

```python
print(df.shape)

df.info()
```

```python
data = df.copy()

print(data.shape)
```

```python
data["CreditScore"] = data["CreditScore"].replace(0, np.nan)

data["CreditScore"] = data["CreditScore"].fillna(
    data["CreditScore"].median()
)

data["CreditScore"].describe()
```

```python
data["SellerName"] = data["SellerName"].fillna(
    data["SellerName"].mode()[0]
)

data["SellerName"].isnull().sum()
```

```python
drop_cols = [
    "LoanSeqNum",
    "MonthsDelinquent"
]

data.drop(columns=drop_cols, inplace=True)

print(data.shape)
```

```python
data["EverDelinquent"].value_counts(normalize=True)
```

```python
data["FirstPaymentDate"] = pd.to_datetime(
    data["FirstPaymentDate"].astype(str),
    format="%Y%m"
)
data["MaturityDate"] = pd.to_datetime(
    data["MaturityDate"].astype(str),
    format="%Y%m"
)
```

```python
data[
    ["FirstPaymentDate", "MaturityDate"]
].head()
```

```python
data["LoanAgeMonths"] = (
    (data["MaturityDate"].dt.year - data["FirstPaymentDate"].dt.year) * 12
    +
    (data["MaturityDate"].dt.month - data["FirstPaymentDate"].dt.month)
)
data["LoanAgeMonths"].describe()
```

```python
data["LoanAgeMonths"].value_counts().sort_index()
```

```python
data["FirstPaymentDate"].value_counts().head(20)
```

```python
data["MaturityDate"].value_counts().head(20)
```

```python
drop_cols = [
    "FirstPaymentDate",
    "MaturityDate",
    "LoanAgeMonths"
]

data.drop(
    columns=drop_cols,
    inplace=True
)

print(data.shape)
```

```python
data.to_csv(
    "../data/processed/freddie_processed.csv",
    index=False
)

print("Dataset saved successfully.")
```
